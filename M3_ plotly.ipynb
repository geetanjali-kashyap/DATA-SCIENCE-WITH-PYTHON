## Bar Chat

```python
from numpy.random.mtrand import uniform

canada_data =  px.data.gapminder().query("country== 'Canada'")

#canada_data = canada_data.sort_values(by = 'pop',ascending = True)
# create a bar chart for canada's population growth over the years
fig = px.bar(canada_data ,x = 'year', y = 'pop',title= "population growth with respect to year in canada" ,labels={ "year ":"years","pop": "population"},hover_data = ['lifeExp','gdpPercap','year','pop'],text = "pop")
# custmize text show on bar.
fig.update_traces(texttemplate = "%{text:.2s}")
# update layout,axis angle,title  and theme
fig.update_layout(xaxis_tickangle = -45, yaxis_title = "population count",template = "plotly_white",uniformtext_minsize = 17,uniformtext_mode = "hide")
# display the final chart
fig.show()
```

## Line Chart

```python
from numpy.random.mtrand import uniform
# create a line chart for canada's population growth over the years
fig = px.line(canada_data ,x = 'year', y = 'pop',title= "population growth with respect to year in canada" ,labels={ "year ":"years","pop": "population"},hover_data = ['lifeExp','gdpPercap','year','pop'],text = "pop")
# custmize text show on bar.
fig.update_traces(texttemplate = "%{text:.2s}")
# update layout.
fig.update_layout(xaxis_tickangle = -45, yaxis_title = "population count",template = "plotly_white",uniformtext_minsize = 17,uniformtext_mode = "hide")

fig.show()
```

## 3d scatter plot

```python
from numpy.random.mtrand import uniform
# create a 3d scatter  chart for canada's population growth over the years
fig = px.scatter_3d(canada_data ,x = 'year', y = 'pop',z = 'gdpPercap',color = 'gdpPercap',size = 'pop',size_max=40,title= "3d population growth with respect to year in canada" )
# update layout.
fig.update_layout(scene = dict(xaxis_title = 'Year', yaxis_title = "population count",zaxis_title= 'GDP per Capita',aspectmode = 'auto'))

fig.show()
```

## Pie Chart

```python
# filter the gapminder dataset to get data only for the year
df_2007 = px.data.gapminder().query("year == 2007")
# create a pie chart showing  population Distribution by continent for 2007
fig = px.pie(df_2007,values = 'pop',names = 'continent',title= "Population Distribution by continent (2007)")
# display the pie chart
fig.show()
```

```python
# load the gapminder dataset
df = px.data.gapminder()
# group data by continent and year, then calculate the average life expectancy
avg_life_exp = df.groupby(["continent","year"],as_index = False)["lifeExp"].mean()
# create multiple chart showing average life EXpectancy trends by continent
fig = px.line(avg_life_exp ,
              x ="year",
              y  = "lifeExp",
              color= "continent",
              markers= True,
              labels= {"year":"Year", "lifeExp": "Avreage Life Expectancy (years)","continent": "Continent"},
              title = "Avreage Life Expentancy by continent(1952 - 2007)")
# custmize font sizes for the chart layout
fig.update_layout(
    title_font_size = 20 ,
    xaxis_title_font_size = 12 ,
    yaxis_title_font_size = 12
)
# display the line chart
fig.show()
```
