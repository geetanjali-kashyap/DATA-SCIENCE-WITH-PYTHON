```python
from string import digits
num = int(input("Enter a number:"))
reverse = 0
while num >0:
    digit = num %10
    reverse = reverse *10+ digit
    num = num// 10
print("reversed number",reverse)
```

```python
num = int(input("Enter a number:"))
original = num
reverse = 0
while num >0:
    digit = num %10
    reverse = reverse *10+ digit
    num = num// 10
if original == reverse:
    print("palindrome number")
else:
    print("it is not a  palindrome number")
```
